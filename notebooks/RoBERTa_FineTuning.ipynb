### 2. Dataset Loading (Kaggle 1.11GB Dataset OR Local Processed Splits)

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from datasets import Dataset, DatasetDict

KAGGLE_PATH_COLAB = "./ai-vs-human-text/AI_Human.csv"
KAGGLE_PATH_KAGGLE = "/kaggle/input/ai-vs-human-text/AI_Human.csv"
LOCAL_PROCESSED_PATH = "../data/processed/train.csv"

# Check if running on Kaggle or Colab with downloaded dataset
if os.path.exists(KAGGLE_PATH_KAGGLE):
    csv_file = KAGGLE_PATH_KAGGLE
elif os.path.exists(KAGGLE_PATH_COLAB):
    csv_file = KAGGLE_PATH_COLAB
elif os.path.exists("../data/raw/AI_Human.csv"):
    csv_file = "../data/raw/AI_Human.csv"
else:
    csv_file = None

if csv_file:
    print(f"Loading Kaggle dataset from: {csv_file}")
    df_full = pd.read_csv(csv_file)
    df_full["label"] = df_full["generated"].astype(int)
    print(f"Total rows in Kaggle dataset: {len(df_full):,}")
    
    # Pre-planned budget: 25,000 to 30,000 balanced rows (fp16 on T4 GPU)
    SAMPLE_SIZE = 25000
    h = df_full[df_full["label"] == 0].sample(SAMPLE_SIZE // 2, random_state=42)
    a = df_full[df_full["label"] == 1].sample(SAMPLE_SIZE // 2, random_state=42)
    df = pd.concat([h, a]).sample(frac=1.0, random_state=42).reset_index(drop=True)
    
    train_df, temp_df = train_test_split(df, test_size=0.30, random_state=42, stratify=df["label"])
    val_df, test_df = train_test_split(temp_df, test_size=0.50, random_state=42, stratify=temp_df["label"])
else:
    print("Loading existing pre-split datasets...")
    train_df = pd.read_csv(LOCAL_PROCESSED_PATH)
    val_df = pd.read_csv("../data/processed/val.csv")
    test_df = pd.read_csv("../data/processed/test.csv")

print(f"Final Splits -> Train: {len(train_df):,} | Val: {len(val_df):,} | Test: {len(test_df):,}")

train_dataset = Dataset.from_pandas(train_df[["text", "label"]].reset_index(drop=True))
val_dataset = Dataset.from_pandas(val_df[["text", "label"]].reset_index(drop=True))
test_dataset = Dataset.from_pandas(test_df[["text", "label"]].reset_index(drop=True))

dataset_dict = DatasetDict({
    "train": train_dataset,
    "validation": val_dataset,
    "test": test_dataset
})
print(dataset_dict)
```

```python
# 3. Tokenization
from transformers import AutoTokenizer

MODEL_CHECKPOINT = "roberta-base"
tokenizer = AutoTokenizer.from_pretrained(MODEL_CHECKPOINT)

def preprocess_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=256,
        padding="max_length"
    )

tokenized_datasets = dataset_dict.map(preprocess_function, batched=True, batch_size=500)
print("Tokenization complete.")
```

```python
# 4. Metrics & Evaluation Function
import numpy as np
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, roc_auc_score

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    probs = torch.softmax(torch.tensor(logits), dim=-1)[:, 1].numpy()
    
    acc = accuracy_score(labels, predictions)
    prec, rec, f1, _ = precision_recall_fscore_support(labels, predictions, average="binary", zero_division=0)
    try:
        auc = roc_auc_score(labels, probs)
    except Exception:
        auc = 0.50
        
    return {
        "accuracy": acc,
        "precision": prec,
        "recall": rec,
        "f1": f1,
        "roc_auc": auc
    }
```

```python
# 5. Model Initialization & Training Arguments
from transformers import AutoModelForSequenceClassification, TrainingArguments, Trainer

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_CHECKPOINT,
    num_labels=2,
    id2label={0: "Human", 1: "AI"},
    label2id={"Human": 0, "AI": 1}
)

training_args = TrainingArguments(
    output_dir="./results_roberta",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    num_train_epochs=2,
    weight_decay=0.01,
    load_best_model_at_end=True,
    metric_for_best_model="f1",
    logging_steps=50,
    fp16=torch.cuda.is_available(),
    seed=42
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["validation"],
    tokenizer=tokenizer,
    compute_metrics=compute_metrics
)

# 6. Fine-Tuning Execution
trainer.train()
```

```python
# 7. In-Domain Test Set Evaluation
test_results = trainer.evaluate(tokenized_datasets["test"])
print("=== In-Domain Test Evaluation Results ===")
for k, v in test_results.items():
    print(f"{k}: {v}")
```

```python
# 8. Save Fine-Tuned Model Weights
SAVE_PATH = "../models/roberta_finetuned"
model.save_pretrained(SAVE_PATH)
tokenizer.save_pretrained(SAVE_PATH)
print(f"Fine-tuned RoBERTa model successfully saved to {SAVE_PATH}")
```
